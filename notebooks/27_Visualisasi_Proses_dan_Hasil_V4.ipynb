```python
import json
import pandas as pd
from IPython.display import display, Image
REPORTS = runner.run_dir / "reports"
print(json.dumps(json.loads((REPORTS / "status.json").read_text()), indent=2))
for name in ["native_diagnostic_summary.csv", "cv_summary.csv", "paired_pooling_deltas.csv", "selected_stability_summary.csv", "stability_within_each_fold.csv", "selected_oof_metrics.csv", "historical_test.csv"]:
    path = REPORTS / name
    if path.exists() and path.stat().st_size > 2:
        print(name)
        display(pd.read_csv(path))
    else:
        print("Pending:", name)
```
